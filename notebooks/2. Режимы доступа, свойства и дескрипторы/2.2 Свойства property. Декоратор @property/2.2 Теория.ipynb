```python
class Person:

    def __init__(self, name, old):
        self.__name = name
        self.__old = old

    def get_old(self):
        return self.__old

    def set_old(self, old):
        self.__old = old

    old = property(get_old, set_old)

# через атрибут old, который является объектом property (видимо это какой-то специальный объект python) можно обращатсья к сеттеру и геттеру, при этом в зависимости от контекста он сам понимает, идет считывание или запись
p = Person('sergey', 20)
p.old = 123
p.old
# если бы old не был property, то в экземпляре класса бы создалось свойство old, но у property приоритет, поэтому происходит так как происходит
```

На данный момент у нас возникло 2 канала взаимодействия с приватными атрибутами: через меоды и через property. Можно сделать лучше:

```python
class Person:

    def __init__(self, name, old):
        self.__name = name
        self.__old = old

    def get_old(self):
        return self.__old

    def set_old(self, old):
        self.__old = old

    old = property()
    old = old.getter(set_old)
    old = old.setter(get_old)

# у property есть 3 декоратора: getter, setter и deleter, так что можно немного переписать
```

Но можно наконец-то написать это нормально через декоратор @property

```python
class Person:

    def __init__(self, name, old):
        self.__name = name
        self.__old = old

    @property # декоратор ставится обязательно перед геттером
    def get_old(self):
        return self.__old

    def set_old(self, old):
        self.__old = old

p = Person('sergey', 20)
p.get_old
```

Но переопределить приватный атрибут через сеттер мы пока не можем, сделаем это далее

```python
class Person:

    def __init__(self, name, old):
        self.__name = name
        self.__old = old

    @property # так get_old становится объектом property, а значит у него появляются присущие ему методы-декораторы
    def old(self):
        return self.__old

    @old.setter # это определяем как сеттер объекта property
    def old(self, old): # этот метод обязательно переименовать, чтоб назывался как исходный property
        self.__old = old

    @old.deleter # удаляет свойство __old.
    def old(self):
        del self.__old

p = Person('sergey', 20)
p.old
p.old = 321
print(p.__dict__)
del p.old
print(p.__dict__)
```
