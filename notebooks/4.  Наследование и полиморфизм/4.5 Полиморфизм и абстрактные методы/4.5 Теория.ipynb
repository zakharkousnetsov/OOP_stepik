Полиморфизм - возможность работать с разными объектами через единый интерфейс

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def get_rect_pr(self):
        return 2*(self.w+self.h)


class Square:
    def __init__(self, a):
        self.a = a

    def get_sq_pr(self):
        return 4*self.a

r1 = Rectangle(1, 2)
r2 = Rectangle(3, 4)

s1 = Square(10)
s2 = Square(20)

print(r1.get_rect_pr(), r2.get_rect_pr())
print(s1.get_sq_pr(), s2.get_sq_pr())
```

все работает, но что если нам бы хотелось перечислить все объекты в одном списке, а потом последовательно для каждого вызвать метод для получения периметра. Но в данном случае имена методов отличаются. Можно ввести ветвление и вызывать соответствующий метод в зависимости от имени класса, но с увеличением числа классов такое ветвление превратится в лапшу. Чтобы добвить гибкости, используют полиморфизм

Договариваемся, что в каждом классе метод, отвечающий за периметр, в каждом классе будет называться одинаково

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def get_pr(self):
        return 2*(self.w+self.h)

class Square:
    def __init__(self, a):
        self.a = a

    def get_pr(self):
        return 4*self.a

class Triange:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def get_pr(self):
        return self.a + self.b + self.c

r1 = Rectangle(1, 2)
r2 = Rectangle(3, 4)

s1 = Square(10)
s2 = Square(20)

tr1 = Triange(1,2,3)
tr2 = Triange(11,11,15)

for obj in (r1, r2, s1, s2, tr1, tr2):
    print(obj.get_pr())
```

У такой реализации есть изъян. Если мы забудем реализовать метод в одном из классов, все поломается. Можно создать базовый класс для всех примитивов, который возвращал бы хотя бы что-то, а потом в каждом отдельном классе мы бы переопределяли этот метод


```python
class Geom:
    def get_pr(self):
        return -1


class Rectangle(Geom):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def get_pr(self):
        return 2*(self.w+self.h)

class Square(Geom):
    def __init__(self, a):
        self.a = a

    def get_pr(self):
        return 4*self.a

class Triange(Geom):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    # def get_pr(self):
    #     return self.a + self.b + self.c

r1 = Rectangle(1, 2)
r2 = Rectangle(3, 4)

s1 = Square(10)
s2 = Square(20)

tr1 = Triange(1,2,3)
tr2 = Triange(11,11,15)

for obj in (r1, r2, s1, s2, tr1, tr2):
    print(obj.get_pr())
```

Но и это не лучшее решение. Можно ввести специальное исключение

```python
class Geom:
    def get_pr(self):
        raise NotImplementedError('в дочернем классе не переопределен метод get_pr')


class Rectangle(Geom):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def get_pr(self):
        return 2*(self.w+self.h)

class Square(Geom):
    def __init__(self, a):
        self.a = a

    def get_pr(self):
        return 4*self.a

class Triange(Geom):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    # def get_pr(self):
    #     return self.a + self.b + self.c

r1 = Rectangle(1, 2)
r2 = Rectangle(3, 4)

s1 = Square(10)
s2 = Square(20)

tr1 = Triange(1,2,3)
tr2 = Triange(11,11,15)

for obj in (r1, r2, s1, s2, tr1, tr2):
    print(obj.get_pr())
```

Методы, которые обязательно должны быть переопределены в дочерних классах, называются абстрактными
