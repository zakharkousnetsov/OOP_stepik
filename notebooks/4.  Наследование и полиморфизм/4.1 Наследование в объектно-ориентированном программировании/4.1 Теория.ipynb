Метод draw не имеет никакого отношения к классу Geom. Но при необходимости мы можем установить связь между классами

```python
class Geom:
    name = 'geom'

class Line(Geom):
    def draw(self):
        print('рисование линии')

g = Geom()
l = Line()

print(l.name)
print(l.__dict__)
```

Когда один класс определяется на основе другого, называется наследование.
Здесь geom базовый или родительский класс,
а Line - подкласс или дочерний класс

```python
class Geom:
    name = 'geom'

class Line(Geom):
    def draw(self):
        print('рисование линии')

    def set_coords(self, x1, y1, x2, y2):
        self.x1 = x1
        self.y1 = y1
        self.x2 = x2
        self.y2 = y2

class Rect(Geom):
    def draw(self):
        print('рисование прямоугоьника')

    def set_coords(self, x1, y1, x2, y2):
        self.x1 = x1
        self.y1 = y1
        self.x2 = x2
        self.y2 = y2

# В данном примере метод set_coords полностью дублируется в двух классах. Лучше общее вынести в базовый класс
```

```python
class Geom:
    name = 'geom'

    def set_coords(self, x1, y1, x2, y2):
        self.x1 = x1
        self.y1 = y1
        self.x2 = x2
        self.y2 = y2

class Line(Geom):
    def draw(self):
        print('рисование линии')

class Rect(Geom):
    def draw(self):
        print('рисование прямоугоьника')

l = Line()
r = Rect()

l.set_coords(1,2,3,4)
r.set_coords(5,6,7,8)


# Метод set_coords сначала ищется в классе Line, и если он там не находится, то поиск продолжается в базовом классе. В таком случае парамер self в def set_coords(self...) ссылается на дочерний класс. То есть когда self вызывается в методе базового класса, он ссылается на экземпляр базового класса, а когда в методе дочернего класса, то ссылается на экземпляр дочернего. Ну и логично, что в случае вызова метода через дочерний класс, все атрибуты в рамках функциональности этого метода присваиваются дочернему классу
```

```python
# теоретически мы можем вызвать метод дочернего класса через базовый

class Geom:
    name = 'geom'

    def set_coords(self, x1, y1, x2, y2):
        self.x1 = x1
        self.y1 = y1
        self.x2 = x2
        self.y2 = y2
        self.draw() # так как self может ссылаться и на базовый класс, и на дочерние, мы через него можем вызывать меоды и базового класса, и дочерних. Но в базовом классе нет методов дочерних классов!

class Line(Geom):
    def draw(self):
        print('рисование линии')

class Rect(Geom):
    def draw(self):
        print('рисование прямоугоьника')

l = Line()
r = Rect()
g = Geom()

l.set_coords(1,2,3,4)
r.set_coords(5,6,7,8)
g.set_coords(5,6,7,8)
```

```python
class Table:
    def __init__(self, model, color):
        self.model = model
        self.color = color


class RoundTable(Table):
    def __init__(self, model, color, radius, height):
        super().__init__(model, color)
        self.radius = radius
        self.height = height

rt = RoundTable(1,2,3,4)
rt.__dict__
```
