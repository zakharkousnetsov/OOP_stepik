Пользовательский класс по умолчанию наследуется от базового класса object

```python
class Geom:
    pass

Geom. # IDE подсказывает методы, которые прописаны в базовом классе object
```

То есть фактически мы могли бы записать вот так. До Пайтон 3 так и писали

```python
class Geom(object):
    pass

g = Geom()
g
```

Вывод выше связан с тем, как написан класс str в базовом классе object

Если ввести новый класс и уже унаследовать его от Geom, то сохранится вся функциональность предыдущих классов

```python
class Line(Geom):
    pass
l = Line()
l
```

Мы можем определить, является ли класс подклассом другого класса

```python
print(issubclass(Line, Geom))
print(issubclass(Line, object))
print(issubclass(Geom, Line)) # экземпляры сюда подставлять нельзя
```

Если мы хотим проверить принадлежность экземпляра классу (или базовому классу), можно использовать isinstance

```python
g = Geom()
l = Line()
print(isinstance(l, Line))
print(isinstance(l, Geom))
print(isinstance(l, object))

print(isinstance(g, Line))
```

```python
print(isinstance(Line, Geom))
print(isinstance(Geom, object))
```

Все стандартные типа данных питона являются классами

```python
print(isinstance(int, object))
```

Раз все стандартные типы являются классами, что нам мешает их расширить?

```python
class Vector(list):
    def __str__(self):
        return '-'.join(map(str, self))

v = Vector([1,2,3])
print(v)
```
