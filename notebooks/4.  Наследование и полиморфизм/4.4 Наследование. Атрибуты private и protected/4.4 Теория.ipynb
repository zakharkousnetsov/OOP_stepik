Разбираемся, как наследуются атрибуты с одним и двумя подчеркиваниями

```python
class Geom:
    name = 'geom'
    def __init__(self, x1, y1, x2, y2):
        print('init Geom')
        self.__x1 = x1
        self.__y1 = y1
        self.__x2 = x2
        self.__y2 = y2

    def get_coords(self):
        '''вот отсюда вызывать метод с приватными атрибутами можно '''
        return (self.__x1, self.__y1)


class Rect(Geom):
    def __init__(self, x1, y1, x2, y2, fill='red'):
        super().__init__(x1, y1, x2, y2)
        self.__fill = fill

    # def get_coords(self):
    #     '''так нельзя!! так как атрибуты имеют название типа _Geom__x1, а не _Rect__x1. Метод нужно перенести в базовый класс и вызывать оттуда'''
    #     return (self.__x1, self.__y1)


r = Rect(0, 1,2,3)
print(r.__dict__)
```

Несмотря на то, что метод вызван для атрибута класса Rect, у всех приватных атрибутов появился префикс базового класса. Кроме _Rect__fill - оно было создано в классе Rect
Мы не можем обращаться к приватным свойства из дочернего класса

Если мы хотел иметь возможность использвоать атрибуты и в дочерних классах. они должны быть protected

```python
class Geom:
    name = 'geom'
    def __init__(self, x1, y1, x2, y2):
        print('init Geom')
        self._x1 = x1
        self._y1 = y1
        self._x2 = x2
        self._y2 = y2

    # def get_coords(self):
    #     '''вот отсюда вызывать метод с приватными атрибутами можно '''
    #     return (self._x1, self._y1)


class Rect(Geom):
    def __init__(self, x1, y1, x2, y2, fill='red'):
        super().__init__(x1, y1, x2, y2)
        self._fill = fill

    def get_coords(self):
        return (self._x1, self._y1)


r = Rect(0, 1,2,3)
print(r.__dict__)
r.get_coords()
```
