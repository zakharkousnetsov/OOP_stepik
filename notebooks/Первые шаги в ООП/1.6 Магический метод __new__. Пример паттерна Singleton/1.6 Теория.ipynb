Если метод __init__ вызывается сразу после создания класса, то непосредственно перед созданием объекта вызывается метод __new__

```python
class Point:
    def __new__(cls, *args, **kwargs):
        # cls ссылается на класс Point
        print('вызов __new__ для ' + str(cls))


    def __init__(self, x=0, y=0):
        # self ссылается на экземпляр класса Point
        print('вызов __init__ для ' + str(self))
        self.x = x
        self.y = y

pt = Point(1, 2)
print(pt)
```

экземпляр класса у нас не создался, потому что мы в методе __new__ не вернули адрес нового объекта

```python
class Point:
    def __new__(cls, *args, **kwargs):
        # cls ссылается на класс Point
        print('вызов __new__ для ' + str(cls))
        # в питоне все объекты наследуются от базового класса object. super() содержит ссылку на базовый класс
        return super().__new__(cls)

    def __init__(self, x=0, y=0):
        # self ссылается на экземпляр класса Point
        print('вызов __init__ для ' + str(self))
        self.x = x
        self.y = y

pt = Point(1, 2)
```

суть в том, чтобы запретить существование одновременно нескольких экземпляров одного класса. Все последующие экземпляры должны вести на первый созданный

```python
class DataBase:
    __instance = None # ссылка на экземпляр класса. если он заполнен, значит объект существует

    def __new__(cls, *args, **kwargs):
        if cls.__instance is None:
            cls.__instance = super().__new__(cls)
        return cls.__instance

    def __del__(self):
        DataBase.__isinstance = None

    def __init__(self, user, psw, port):
        self.user = user
        self.psw = psw
        self.port = port

    def connect(self):
        print(f'соединение с БД: {self.user}, {self.psw}, {self.port}')

    def close(self):
        print(f'закрытие соединения с БД')

    def read(self):
        print(f'даные из БД')

    def write(self, data):
        print(f'данные в БД {data}')


db1 = DataBase('root', '1234', '80')
db2 = DataBase('root2', '9999', '83450')
db1 == db2
# второй экземпляр ссылается на первый
```

```python
db2.connect()
db2.connect()
# у нас первый объект заполнился данными из второго, потому что фактически метод __init__ был вызван заново. Чтобы это поправить, нужно переопределить метод __call__
```
