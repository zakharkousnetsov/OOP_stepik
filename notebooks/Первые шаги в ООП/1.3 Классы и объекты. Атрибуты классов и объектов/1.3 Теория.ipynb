```python
class Point: # класс начинается  сзаглавной буквы
    color = 'red' # атрибут (свойство) класса
    circle = 2

Point.color
```

```python
# переопределение атрибута класса
Point.color = 'black'
Point.color
```

```python
# увидеть все атрибуты класса
Point.__dict__
```

```python
# создание экземпляра класса
a = Point()
b = Point()

print(type(a))
print(type(a) == Point)
print(isinstance(a, Point))
```

```python
# Атрибуты класса являются общими для объектов a и b
print(a.circle)
print(b.circle)
```

```python
# теперь переопределим один из атрибутов класса
Point.circle = 1
a.circle
```

```python
#  сам по себе объект класса не содержит никаких собственных атрибутов, только атрибуты класса
a.__dict__
```

```python
# если попробовать изменить атрибус класа через объект класса, то ничего не получится, а в пространстве имен объекта создастся новый атрибут, характерный уже только для этого объекта
a.color = 'green'
print(a.__dict__)
print(b.color)
print(Point.color)
```

```python
# можно создавать новые атрибуты класса, и они перенесутся во все объекты
Point.type_pt = 'disk'
print(Point.__dict__)
print(a.__dict__)
print(b.__dict__)
```

```python
# то же самое можно сделать при помощи команды setattr
setattr(Point, 'prop', 1)
print(Point.__dict__)
print(a.__dict__)
print(b.__dict__)
```

```python
# при помощи этой команды можно переопределить заранее заданное свойство класса
setattr(Point, 'type_pt', 'square')
print(Point.__dict__)
```

```python
# прочитать атрибут класса
print(Point.circle)             # если указанного атрибута нет, вернется ошибка
getattr(Point, 'circle', False) # если указанного атрибута нет, вернется False
```

```python
# Удалить атрибут класса. если атрибута не существует, вернется ошибка
del Point.prop
Point.__dict__
```

```python
# проверка, есть ли в классе такой атрибут. Можно также проверить наличие атрибута через экземпляр класса. Короче hasattr проверяет как атрибуты класса, так и пространство имен объекта. 
# Проверить конкретное пространство можно выражением "attr" in <class_name>.__dict__
print(hasattr(Point, 'abcd'))
print(hasattr(Point, 'circle'))
```

```python
# еще удалить атрибут можно командой delattr. если мы попытаемся удалить несуществуюий атрибус, вернется ошибка. Здесь нет возможности указать False, нужно каждый раз проверять наличие атрибута
delattr(Point, 'circle')
delattr(Point, 'circle')
```

```python
# если мы удалим перекликающееся имя из пространства имен экземпляра, то оно будет браться из числа атрибутов класса. Таков порядок обращения к атрибутам: сначала из внутреннего пространства имен, потом из внешнего
print(a.__dict__, a.color)
del a.color
print(a.__dict__, a.color)
```
