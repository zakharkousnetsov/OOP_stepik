Класс - это шаблон объекта с характерными ему свойствами (атрибутами) и функциями (методами)  
Инкапсуляция - возможность разделять методы и атрибуты на внутренние (скрытые) и публичные  
Наследование - возможность использовать созданыне ранее классы и расширять их функциональность  
Полиморфизм - возможность переопределять функциональность базового класса, но это не точно. Универсальность интерфейса доступа к разным типам данных
