Магические методы начинаются и заканчиваются двумя подчеркиваниями. В частности, есть магические методы  
__init__(self)  - инициализатор класса
__del__(self) - финализатор класса

```python
class Point():
    color = 'red'
    circle = 2

    def set_coords(self, x, y):
        self.x = x
        self.y = y

    def get_coords(self):
        return self.x, self.y
```

классом не очень удобно пользоваться. наделить экземпляр класса атрибутами можно только через метод set_coords. Было бы удобно сделать это сразу в момент создания экземпляра класса

```python
class Point():
    color = 'red'
    circle = 2

    def __init__(self, x, y=12): # можно указывать аргументы по умолчанию
        print('вызов init')
        self.x = x
        self.y = y

    def __del__(self):
        print(f'удаление {str(self)}')

    def set_coords(self, x, y):
        self.x = x
        self.y = y

    def get_coords(self):
        return self.x, self.y
```

```python
pt = Point(1,)
# pt.__dict__
```

Метод __del__ срабатывает, когда объект становится ненужным. Это определяет сборщик мусора. Он определяет, ведет ли на объект хотя бы одна ссылка, если да, то объект нужный
