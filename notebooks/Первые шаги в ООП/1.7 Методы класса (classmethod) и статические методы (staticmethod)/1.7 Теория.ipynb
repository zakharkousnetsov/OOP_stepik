 Ранее мы прошли обычные методы. А есть еще статические методы и методы класса



```python
class Vector:
    MIN_COORD = 0
    MAX_COORD = 100

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def get_coord(self):
        return self.x, self.y

v = Vector(1, 2)
Vector.get_coord(v) # чтобы вызвать метод через класс, мы обязаны передать что-то на место self, то есть ссылку на какой-нибудь экземпляр класса
```

```python
class Vector:
    MIN_COORD = 0
    MAX_COORD = 100

    @classmethod # метод класса работает исключительно с атрибутами класса. Но не может работать с локальными атрибутами экземпляров этого класса
    # Ну и так как у этого метода нет ссылки на объект, через него мы не можем ничего делать с экземплярами класса
    def validate(cls, arg):
        return cls.MIN_COORD <= arg <= cls.MAX_COORD

    def __init__(self, x, y):
        self.x = x
        self.y = y

print(Vector.validate(5666)) # а здесь у нас метод класса, для его вызова не нужен никакой экземпляр
# ну и по идее метод все-таки можно вызвать через экземпляр, но опять же никакие локальные свойства экземпляра мы получить или изменить через такой метод не сможем
print(v.validate(5666))
```

```python
class Vector:
    MIN_COORD = 0
    MAX_COORD = 100

    @classmethod # метод класса работает исключительно с атрибутами класса. Но не может работать с локальными атрибутами экземпляров этого класса
    # Ну и так как у этого метода нет ссылки на объект, через него мы не можем ничего делать с экземплярами класса
    def validate(cls, arg):
        return cls.MIN_COORD <= arg <= cls.MAX_COORD

    def __init__(self, x, y):
        self.x = self.y = 0
        # через метод класса можно реализовать проверку. Тут мы второе условие проверяем через параметр self, так как в каждом экземпляре содержится такой метод
        if Vector.validate(x) and self.validate(y):
            self.x = x
            self.y = y

v = Vector(1,2)
v.__dict__
```

Статические методы не имеют доступа ни к атрибутам класса, ни к атрибутам экземпляров. Этот же меод можно вызывать внутри других методов класса

```python
class Vector:
    MIN_COORD = 0
    MAX_COORD = 100

    @classmethod # метод класса работает исключительно с атрибутами класса. Но не может работать с локальными атрибутами экземпляров этого класса
    # Ну и так как у этого метода нет ссылки на объект, через него мы не можем ничего делать с экземплярами класса
    def validate(cls, arg):
        return cls.MIN_COORD <= arg <= cls.MAX_COORD

    def __init__(self, x, y):
        self.x = self.y = 0
        # через метод класса можно реализовать проверку. Тут мы второе условие проверяем через параметр self, так как в каждом экземпляре содержится такой метод
        if Vector.validate(x) and self.validate(y):
            self.x = x
            self.y = y
        print('self.norm2(self.x, self.y) = ', self.norm2(self.x, self.y))

    @staticmethod
    def norm2(x, y):
        # технически сюда можно было бы подставить Vector.MIN_COORD, но это идеологически неправильно
        # это должна быть типа вспомогательная сервисная функция
        return x**2 + y**2

v = Vector(5, 6)
Vector.norm2(5, 6)

```
