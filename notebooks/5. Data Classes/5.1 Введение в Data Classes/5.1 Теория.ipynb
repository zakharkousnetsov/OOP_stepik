Классы данных. Чаще всего при создании класса нужно приписывать инициализатор. К тому же, если вывести на экран экземпляр класса, то по умолчанию методы __str__ и __repr__ демонстируруют нефинформативные характеристики типа ячейки памяти

```python
class Thing:
    def __init__(self, name, weight, price):
        self.name = name
        self.weight = weight
        self.price = price

    def __repr__(self):
        return f'{self.__class__}: {self.__dict__}'

t = Thing('zakhar', 80, 999)
t
```

```python
from dataclasses import dataclass, field
from pprint import pprint

@dataclass
class ThingData:
    name: str
    weight: int
    price: float

td1 = ThingData('zakhar', 80, 999)
td1
```

```python
pprint(ThingData.__dict__)
```

Здесь видно, что в этом классе определено довольно много стандартных методов

```python
td2 = ThingData('zaki', 80, 999)
td1 == td2
```

Вот например в методе __eq__ сравниваются попарно все свойства. Можно переопределить и сравнивать, например, только указанные
