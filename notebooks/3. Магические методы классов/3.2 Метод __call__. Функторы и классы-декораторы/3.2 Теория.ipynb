```python
class Counter:
    def __init__(self):
        self.__counter = 0

c = Counter()
# Когда мы ставим две круглые скобки, мы вызываем класс подобно функции, в этот момент вызывается метод __call__:
# __call_(self, *args, **kwargs):

# При создании экземпляра класса срабатывает такой алгоритм:
# obj = self.__new__(self, *args, **kwargs)
# self.__init__(obj, *args, **kwargs)
# return obj
```

```python
# класс при помощи круглых скобок мы вызывать не можем, потому что для них не определен метод __call__
c()
```

```python
# переопределим метод __call__, чтобы уметь вызывать экземпляры.
# Классы которые имеют такую возможность, называются функторы

class Counter:
    def __init__(self):
        self.__counter = 0

    def __call__(self, *args, **kwargs):
        print('__call__')
        self.__counter += 1
        return self.__counter

c1 = Counter()
c1()
c1()
res1 = c1()

c2 = Counter()
res2 = c2()

print(res1, res2)
# счетчики для экземпляров независимы
```

```python
class Counter:
    def __init__(self):
        self.__counter = 0

    def __call__(self, step=1, *args, **kwargs):
        # задали кастомный шаг счетчика
        print('__call__')
        self.__counter += step
        return self.__counter

c1 = Counter()
c1(1)
c1(12)
```

Этот класс можно использовать вместо замыкания функции

```python
class StripChars:
    def __init__(self, chars):
        self.__counter = 0
        self.__chars = chars

    def __call__(self, *args, **kwargs):
        if not isinstance(args[0], str):
            raise TypeError('аргумент должен быть строкой')

        return args[0].strip(self.__chars)

s1 = StripChars('?:!.; ')
s1('       hello world!   ')

```

С помощью классов можно сделать декоратор

```python
import math

class Derivate:
    def __init__(self, func):
        self.__fn = func

    def __call__(self, x, dx=0.0001, *args, **kwargs):

        return (self.__fn(x+dx)-self.__fn(x))/dx

def df_sin(x):
    return math.sin(x)

print(df_sin(math.pi/3))

# переопределям переменную через новый класс
df_sin = Derivate(df_sin)
print(df_sin(math.pi/3))
```

```python
# А еще можно использовать в чистом виде декоратор

@Derivate
def df_sin(x):
    return math.sin(x)

print(df_sin(math.pi/3))
```
