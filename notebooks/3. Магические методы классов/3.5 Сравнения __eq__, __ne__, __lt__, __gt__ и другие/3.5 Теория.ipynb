```python
# Магические методы сравнения
#
# __eq__() - для равенства ==
# __пе__() - для неравенства !=
# __It__() - для оператора меньше <
# __le__() - для оператора меньше или равно <=
# __gt__() - для оператора больше >
# __ge__() - для оператора больше или равно >=
```

```python
class Clock:
    __DAY = 86400 # число секунд в одном дне

    def __init__(self, seconds: int):
        if not isinstance(seconds, int):
            raise TypeError('секунды должны быть целым числом')
        self.seconds = seconds % self.__DAY

    def __eq__(self, other):
        '''метод работает для операторов == и !='''
        if not isinstance(other, (int, Clock)):
            raise TypeError('int or clock')
        sc = other if isinstance(other, int) else other.seconds
        return self.seconds == sc

    def __lt__(self, other):
        '''метод работает для оператора <, оператор > реализован автоматически, если не переопределен __gt__()'''
        if not isinstance(other, (int, Clock)):
            raise TypeError('int or clock')
        sc = other if isinstance(other, int) else other.seconds
        return self.seconds < sc

    def __le__(self, other):
        '''метод работает для оператора <=, оператор > реализован автоматически, если не переопределен __ge__()'''
        if not isinstance(other, (int, Clock)):
            raise TypeError('int or clock')
        sc = other if isinstance(other, int) else other.seconds
        return self.seconds <= sc

c1 = Clock(1000)
c2 = Clock(1233)
# мы бы хотели иметь возможность сравнивать экземпляры между собой

print('Clock(1000) == Clock(1000)', Clock(1000) == Clock(1000))
print('Clock(1000) == Clock(1234)', Clock(1000) == Clock(1234))

# != определен по умолчанию, если не переопределен метод __ne__()
print('Clock(1000) != Clock(1000)', Clock(1000) != Clock(1000))
print('Clock(1000) != Clock(1234)', Clock(1000) != Clock(1234))

print('Clock(1000) < Clock(1000)', Clock(1000) < Clock(1000))
print('Clock(1000) < Clock(1234)', Clock(1000) < Clock(1234))

print('Clock(1000) > Clock(1000)', Clock(1000) > Clock(1000))
print('Clock(1000) > Clock(1234)', Clock(1000) > Clock(1234))

print('Clock(1000) >= Clock(1000)', Clock(1000) >= Clock(1000))
print('Clock(1000) >= Clock(1234)', Clock(1000) >= Clock(1234))
```
