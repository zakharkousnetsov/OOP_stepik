```python
class WordString:
    def __init__(self, string=''):
        self.string = string

    @property
    def string(self):
        return self.__string

    @string.setter
    def string(self, string):
        self.__string = string

    def __len__(self):
        return len([word for word in self.string.split(' ') if word != ''])

    def __call__(self, indx):
        return [word for word in self.string.split(' ') if word != ''][indx]

w = WordString('zero   first   second third')
print(len(w))
print(w(1))
```

remove_obj(indx) - удаление объекта класса ObjList из связного списка по его порядковому номеру (индексу); индекс отсчитывается с нуля.

```python
class Descr:
    def __set_name__(self, owner, name):
        self.name = f'_{owner.__name__}__{name}'

    def __get__(self, instance, owner):
        return getattr(instance, self.name)

    def __set__(self, instance, value):
        setattr(instance, self.name, value)

class ObjList:

    prev = Descr()
    next = Descr()

    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.counter = 0

    def add_obj(self, obj: ObjList):
        self.counter += 1
        if not self.head:
            self.head = obj
            self.tail = obj
        else:
            current_tail = self.tail
            self.tail.next = obj
            self.tail = obj
            self.tail.prev = current_tail

    def remove_obj(self, indx):
        if not self.head or self.head.next is None: # если элементов 1 или 0, возвращаем ничего
            self.head = None
            self.tail = None
        else:
            ptr = self.head
            i = 0
            while i != indx:
                ptr = ptr.next
                i += 1

            if ptr.prev and ptr.next: # если удаляемый элемент где-то посередине
                ptr.prev.next = ptr.next
                ptr.next.prev = ptr.prev
            elif ptr.prev and not ptr.next: # если он в конце
                self.tail = ptr.prev
                self.tail.next = None
            else: # иначе он в начале
                self.head = ptr.next
            self.counter -= 1


    def __len__(self):
        return self.counter

    def __call__(self, indx):
        if indx <= len(self)-1:
            ptr = self.head
            for i in range(indx):
                ptr = ptr.next
            return ptr.data
        else:
            return None

o1 = ObjList('o1')
o2 = ObjList('o2')
o3 = ObjList('o3')

# o5 = ObjList('o5')

ll = LinkedList()
ll.add_obj(o1)
ll.add_obj(o2)
ll.add_obj(o3)

ll.remove_obj(0)

# o4 = ObjList('o4')
# ll.add_obj(o4)
#
#
print(ll.head.data, ll.head.next.data, ll.tail.data, sep='\n')
```

```python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    @property
    def real(self):
        return self.__real

    @real.setter
    def real(self, value):
        self.__real = value

    @property
    def img(self):
        return self.__img

    @img.setter
    def img(self, value):
        self.__img = value

    def __setattr__(self, key, value):

        if not isinstance(value, (int, float)):
            raise ValueError("Неверный тип данных.")
        object.__setattr__(self, key, value)

    def __abs__(self):
        return (self.real*self.real + self.img*self.img)**0.5

cmp = Complex(7, 8)
cmp.real, cmp.img = 3, 4
c_abs = abs(cmp)
c_abs
```

Класс RadiusVector должен содержать методы:

```python
class RadiusVector:
    def __init__(self, *args):

        if len(args) == 1 and isinstance(args[0], int) and args[0]>1:
            self.coords = [0]*args[0]
        else:
            self.coords = [*args]

    def set_coords(self, *args):
        for i in range(min(len(args), len(self.coords))):
            self.coords[i] = args[i]

    def get_coords(self):
        return self.coords

    def __len__(self):
        return len(self.coords)

    def __abs__(self):
        return sum([coord**2 for coord in self.coords])**0.5


rv = RadiusVector(4, 5, 6, 7)
# rv.set_coords(1, 2, 6,7,8,6)
abs(rv)
```

```python
class Clock:
    def __init__(self, hours, minutes, seconds):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds

    def get_time(self):
        return self.hours*60*60 + self.minutes*60 + self.seconds

    def __setattr__(self, key, value):
        if not isinstance(value, int) or value<0:
            raise TypeError('плохо-плохо-плохо-плохо')
        super().__setattr__(key, value)

class DeltaClock:
    def __init__(self, clock1: Clock , clock2: Clock):
        self.clock1 = clock1
        self.clock2 = clock2
        self.inverse_transform()

    def inverse_transform(self):
        self.sec_diff = self.clock1.get_time() - self.clock2.get_time()
        if self.sec_diff<0:
            self.hours, self.minutes, self.seconds = 0, 0, 0
        else:
            self.hours = self.sec_diff // (60*60)
            self.minutes = (self.sec_diff-self.hours*60*60) // 60
            self.seconds = (self.sec_diff-(self.hours*60*60)-(self.minutes*60)) % 60

    def __str__(self):

        return f'{self.hours:02d}: {self.minutes:02d}: {self.seconds:02d}'

    def __len__(self):
        return max(0, self.sec_diff)

c1 = Clock(15, 22, 11)
c2 = Clock(4, 2, 5)
dt = DeltaClock(c1, c2)
str(dt)
```

С объектами класса Ingredient должна работать функция:

str(ing)  # название: объем, ед. изм.

Класс Recipe должен иметь следующие методы:

add_ingredient(ing) - добавление нового ингредиента ing (объект класса Ingredient) в рецепт (в конец);
remove_ingredient(ing) - удаление ингредиента по объекту ing (объект класса Ingredient) из рецепта;
get_ingredients() - получение кортежа из объектов класса Ingredient текущего рецепта.

Также с объектами класса Recipe должна поддерживаться функция:

len(recipe) - возвращает число ингредиентов в рецепте.

Пример использования классов (эти строчки в программе писать не нужно):

recipe = Recipe()

ings = recipe.get_ingredients()

```python
class Ingredient:
    def __init__(self, name: str, volume: float, measure: str):
        self.name = name
        self.volume = volume
        self.measure = measure

    def __str__(self):
        return f'{self.name}: {self.volume}, {self.measure}'


class Recipe:
    def __init__(self, *args: Ingredient):
        self.ings = list(args)

    def add_ingredient(self, ing):
        self.ings.append(ing)

    def remove_ingredient(self, ing):
        self.ings.remove(ing)

    def get_ingredients(self):
        return tuple(self.ings)

    def __len__(self):
        return len(self.ings)

recipe = Recipe()
recipe.add_ingredient(Ingredient("Соль", 1, "столовая ложка"))
recipe.add_ingredient(Ingredient("Мука", 1, "кг"))
recipe.add_ingredient(Ingredient("Мясо баранины", 10, "кг"))
ings = recipe.get_ingredients()
len(recipe) # n = 3
```
