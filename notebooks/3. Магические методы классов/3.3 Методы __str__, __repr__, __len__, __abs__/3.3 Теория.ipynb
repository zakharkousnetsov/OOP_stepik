```python
# __str__() - для отображения информации об объекте класса для пользователей (например, для функций print, str);
# __repr__() - для отображения информации об объекте класса в режиме отладки (для разработчиков).
```

```python
class Cat:
    def __init__(self, name):
        self.name = name

cat = Cat('vaska')
cat
# по умолчанию мы видим имя класса и адрес, где хранится экземпляр
```

При помощи методов str и repr можно изменить отображение информации об экземпляре

```python
class Cat:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f'{self.__class__}: {self.name}'

cat = Cat('vaska')
print(cat)
# это уже скорее вывод данных для пользователя, тут должен отрабатывать метод __str__,
# а не отрабатывает он потому, что мы его не определили

cat
# __repr__ вызывается каждый раз, когда идет как бы вызов экземпляра (не всмысле вызов как с круглыми скобками)
```

```python
class Cat:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f'{self.__class__}: {self.name}'

    def __str__(self):
        return f'__str__: {self.name}'

cat = Cat('vaska')
print(cat)
# это уже скорее вывод данных для пользователя, тут должен отрабатывать метод __str__,
# а не отрабатывает он потому, что мы его не определили

cat
# __repr__ вызывается каждый раз, когда идет как бы вызов экземпляра (не всмысле вызов как с круглыми скобками)
```

```python
# __len__() - позволяет применять функцию len() к экземплярам класса (к указанным атрибутам);
# __abs__() - позволяет применять функцию abs() к экземплярам класса.
```

```python
class Point:
    def __init__(self, *args):
        self.__coords = args

p = Point(1, 2)
len(p)
```

Возникает ошибка, а мы бы хотели, чтобы функция len возвращала длину какого-нибудь атрибута

```python
class Point:
    def __init__(self, *args):
        self.__coords = args

    def __len__(self):
        return len(self.__coords)

p = Point(1, 2)
len(p)
```

Следующим образом работает метод __abs__

```python
class Point:
    def __init__(self, *args):
        self.__coords = args

    def __abs__(self):
        return list(map(abs, self.__coords))

p = Point(1, -2)
abs(p)
```
