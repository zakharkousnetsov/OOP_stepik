Пространство атрибутов класса является общим для всех экземпляров.
Обращатсья внутри методов напрямую к атрибутам класса нельзя, только через имя класса или self.

```python
class Point():
    MIN_COORD = 0
    MAX_COORD = 100

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def set_coord(self, x, y):
        self.x = x
        self.y = y

    def set_bound(self, left):
        self.MIN_COORD = left
#         так делать нельзя, так создается атрибут внутри экземпляра класса, а атрибут класса не изменяется
#         Правильней было бы использовать @classmethod

pt = Point(1, 2)
pt.set_bound(-100)
print(pt.__dict__)
print(Point.__dict__)
```

Магические методы для атрибутов (по 2 нижних подчеркивания слева и справа)
•__setattr__(self, key, value) — автоматически вызывается при изменении свойства кеу класса;
•__getattribute__(self, item) - автоматически вызывается при получении свойства класса с именем item;
•__getattr__(self, item) - автоматически вызывается при получении несуществующего свойства item класса;
•__delattr__(self, item) - автоматически вызывается при удалении свойства item (не важно: существует оно или нет).

# __getattribute__

```python
class Point():
    MIN_COORD = 0
    MAX_COORD = 100

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def set_coord(self, x, y):
        self.x = x
        self.y = y

    def __getattribute__(self, item):
        '''вызывается автоматически, когда идет считывание через экземпляр класса
        при помощи такого метод, например, можно запретить обращатсья к какому-то атрибуту'''

        if item == 'y':
            raise ValueError('доступ запрещен')
        print('вызван __getattribute__')
        return object.__getattribute__(self, item)


pt = Point(1, 2)
pt.x
pt.y
```

# __setattr__

```python
class Point():
    MIN_COORD = 0
    MAX_COORD = 100

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def set_coord(self, x, y):
        self.x = x
        self.y = y

    def __setattr__(self, key, value):
        '''вызывается, когда идет присвоение значения какому-либо атрибуту
        с посощью данного метода мы можем запретить создавать создавать атрибут с конкретным именем'''
        if key == 'z':
            raise ValueError('такое имя запрещено')

        print('вызван __setattr__')
        object.__setattr__(self, key, value)
        # self.s__dict__[key] = value # так можно, но лучше через object
        # self.key = value # можно было бы написать так, но тогда для объекта self вызовется метод __setattr__ и возникнет рекурсия

pt = Point(1, 2)
pt.z = 3
```

# delattr

```python
class Point():
    MIN_COORD = 0
    MAX_COORD = 100

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def set_coord(self, x, y):
        self.x = x
        self.y = y

    def __delattr__(self, item):
        print(f'__delattr__ удален {item}')
        object.__delattr__(self, item)


pt = Point(1, 2)
print(pt.__dict__)
del pt.x
print(pt.__dict__)
```
