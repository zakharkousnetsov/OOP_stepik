В питоне есть возможность вычислять некий индекс для неизменяемых объектов, который всегда будет одинаковым для одинаковых объектов.
Но одинаковым хешам иногда могут соответствовать разные объекты.
если хеши не равны, то объекты точно не равны
Хеш можно считать только для неизменяемого объекта, например для списка нельзя

```python
hash(3)
```

```python
hash('abc')
```

```python
hash((1,2,3))
```

Для ключей словаря можно использовать только хешируемые объекты. Когда мы сохраняем нвое значение в словаре по ключу, словарь хрнит ключ в формате кортежа (хеш, ключ). Это нужно для быстрого поиска, реализованного как раз через хеш

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = Point(1, 2)
p2 = Point(1, 2)

print(hash(p1), hash(p1), hash(p2), sep='\n')
#  как хеш определяет, разные объекты или одинеаковые? Если obj1==obj2, то хеши равны
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x==other.x and self.y==other.y

p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1==p2)
print(hash(p1))  # ошибка!
# Ошибка в том, что мы перепоеределили метод __eq__, и функция hash перестала работать. Чтобы функция снова зарабаотала, нужно в класе объявить метлд __hash__
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x==other.x and self.y==other.y

    def __hash__(self):
        return hash((self.x, self.y))

p1 = Point(1, 2)
p2 = Point(1, 2)


print(hash(p1), hash(p2), sep='\n')  # ошибка!
```

```python
# В результате если мы попытаемся сделать эти экземпляры ключами словаря, то они будут восприниматься как один ключ, так как их хеши равны
dct = {p1: 1, p2: 2}
dct
```

В итоге, если мы хотим, чтобы экземпляры с некоторыми одинаковыми свойствами воспринимались как один объект, можно переопределеить метод __hash__
