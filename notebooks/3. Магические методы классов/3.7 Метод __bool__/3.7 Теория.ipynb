Способы определения правдивости класса, это означает, что для экземпляра становится возможным применением булевых операций

```python
# __len__() - вызывается функцией bool(), если не определен магический метод_bool__();
# __bool__() - вызывается в приоритетном порядке функцией bool().
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(3, 4)
print(bool(p))
# такая операция всегда возвращает True для любых объектов пользовательского класса
# Но мы можем переопределить ее поведение через магические методы __len__ и __bool__
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __len__(self):
        print('len called')
        return self.x**2 + self.y**2

p = Point(3, 4)
print(len(p))
print(bool(p))
# получили True, так как bool от любого ненулевого числа возвращает True
print(bool(Point(0, 0)))
```

если нам нужно как-то явно переопределить метод bool, то можно сделать вот так

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __len__(self):
        print('len called')
        return self.x**2 + self.y**2

    def __bool__(self): # у метода bool приоритет по отношению к len
        return self.x==self.y # должен возвращать True/False

p = Point(42, 4)
print(bool(p))
```

```python
p = Point(42, 42)
if p:
    print('координаты равны')
else:
    print('координаты НЕ равны')
```
