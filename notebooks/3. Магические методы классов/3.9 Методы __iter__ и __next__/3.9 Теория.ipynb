```python
# Магические методы __iter__ И __next__
# __iter__(self) - получение итератора для перебора объекта;
# __next__(self) - переход к следующему значению и его считывание.
```

```python
class Frange:
    def __init__(self, start=0.0, stop=0.0, step=1.0):
        self.start = start
        self.stop = stop
        self.step = step
        self.value = self.start - self.step

    def __next__(self): # переопределили метод, чтобы использовать next
        if self.value + self.step < self.stop:
            self.value += self.step
            return self.value
        else:
            raise StopIteration

fr = Frange(0, 2, 0.5) # объект выступает итератором
print(fr.__next__())
print(fr.__next__())
print(fr.__next__())
print(next(fr))
```

```python
for x in fr:
    print(fr)

# ошибка, так как мы не можем создать итератор для этого объекта, то есть iter(fr)
```
