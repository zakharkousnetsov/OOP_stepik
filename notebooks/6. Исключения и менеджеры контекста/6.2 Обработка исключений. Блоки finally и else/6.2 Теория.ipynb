блоков try/except бывает недостаточно. Бывает, приходится прпоисывать блоки finally/else

```python
try:
    x, y = map(int, input().split())
    res = x/y
except ZeroDivisionError as z:
    print(z)
except ValueError as z:
    print(z)
else: # выполняется при штатном выполении кода в блоке try
    print('все отработало')
finally:
    print('finally выполняется всегда')

```

когда же может понадобитсья блок finally? Например при работе с файлом. Хотя более корректно было бы использовать оператор with

```python
try:
    f = open('file.txt') # файл открывается на чтение
    f.write('some text') # а тут мы пытаемся в него что-то записать. Возникает ошибка ValueError, а файл остается открытым
except FileNotFoundError as z:
    print(z)
except ValueError:
    print('все остальные ошибки')
finally:
    if f: # если файл открыт
        f.close() # закрываем файл

```

Есть особенность в последовательности выполнения операций

```python
def get_values():
    try:
        x, y = map(int, input().split())
        return x, y
    except ValueError as err:
        print(err)
        return 0, 0
    finally:
        print('finally выполняется до return')
x, y = get_values()
```

В юпитер ноутбуке (да и в консоли) не особо видно, но типо return выполняется в последнюю очередь


Блоки try/except можно использовать вложенно

```python
try:
    x, y = map(int, input().split())
    try:
        res = x/y
    except ZeroDivisionError as err:
        print('деление на ноль')
except ValueError as err:
    print(err)
```
