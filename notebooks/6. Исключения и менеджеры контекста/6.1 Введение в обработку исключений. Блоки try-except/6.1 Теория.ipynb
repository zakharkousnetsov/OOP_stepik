Исключения бывают двух видов:
* Исключения при выполнении (ошибка типа, ошибка имени, деление на ноль и тд)
* Исключения при компиляции (когда код вообще не запустился)

Мы хотим отлавливать только первый тип

```python
try:
    f = open('file.txt')
except FileNotFoundError: # указываем, в случае какого типа ошибки можно продолжать работу FileNotFoundError. Можно ничего не указывать, тогда будет пропускаться всё
    print('файл не найден')

print('работаем дальше')
```

```python
try:
    f = open('file.txt')
except ValueError: # если случится ошибка не та, которую мы отлавливаем, программа упадет
    print('файл не найден')

print('работаем дальше')
```

Что делать, если мы хотим отлавливать сразу несколько типов ошибок?

```python
try:
    f = open('file.txt')
    res = 1/0
except FileNotFoundError: # если случится ошибка не та, которую мы отлавливаем, программа упадет
    print('файл не найден')

except ZeroDivisionError: # если случится ошибка не та, которую мы отлавливаем, программа упадет
    print('деление на ноль')

print('работаем дальше')
```

Можно ловить за раз несколько исключений

```python
try:
    f = open('file.txt')
    res = 1/0
except (FileNotFoundError, ZeroDivisionError):
    print('файл не найден или деление на ноль')

print('работаем дальше')
```

Но удобней все-таки каждое исключение обрабатывать отдельно через множественный except

Исключения реализованы через классы со своей иерарархией. Самый главный класс - Exception, от него наследуются остальные. Соответственно если указать просто Exception, то будут отловлены все его дочерние классы

- `BaseException`
  - `Exception`
    - `AttributeError`
    - `ArithmeticError`
      - `ZeroDivisionError`
      - `FloatingPointError`
      - `OverflowError`
    - `EOFError`
    - `NameError`
    - `LookupError`
      - `IndexError`
      - `KeyError`
    - `OSError`
      - `FileNotFoundError`
      - `InterruptedError`
      - `PermissionError`
      - `TimeoutError`
    - `TypeError`
    - `ValueError`
  - `SystemExit`
  - `GeneratorExit`
  - `KeyboardInterrupt`


```python
try:
    res = 1/0
except ArithmeticError:
    print('файл не найден или деление на ноль')

print('работаем дальше')
```

Учитывая иерархию, обычно принято первыми прописывать блоки с узок-специализированными исключениями, а уже в конце ставить просто Exception, который ловит "все неуказанное"

```python
try:
    res = 1/0
except FileNotFoundError:
    print('файл не найден')

except Exception:
# except # можно написть просто except
    print('общее исключение')

print('работаем дальше')
```

```python
def get_number(x):
    try:
        return int(x)
    except:
        try:
            return float(x)
        except:
            return x


res_1 = get_number('-5')
res_2 = get_number('5.78')
res_3 = get_number('8(912)000-000-00')
```
