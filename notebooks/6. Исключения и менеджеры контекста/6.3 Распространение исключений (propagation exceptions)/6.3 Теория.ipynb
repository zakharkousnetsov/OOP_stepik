ранее были простые примерчики, теперь посмотрим, как писать обработку ошибку на примере вызова функции

```python
def func2():
    return 1/0

def func1():
    func2()


print('1')
print('2')
print('3')
func1()
print('4')
print('5')
print('6')


```

Когда вызывается функция, она помещается в стек вызовов функции. Соответственно исключение возникло на этапе func2, перешло на func1 и распространилось на main в строчке 11

Исключения можно обрабатывать на любом эатпе, хоть в самой функции, хоть при ее вызове

```python
def func2():
    return 1/0

def func1():
    func2()

print('1')
print('2')
print('3')
try:
    func1()
except Exception as err:
    print(err)

print('4')
print('5')
print('6')


```

```python
def func2():
    return 1/0

def func1():
    try:
        func2()
    except Exception as err:
        print(err)

print('1')
print('2')
print('3')
func1()

print('4')
print('5')
print('6')


```
