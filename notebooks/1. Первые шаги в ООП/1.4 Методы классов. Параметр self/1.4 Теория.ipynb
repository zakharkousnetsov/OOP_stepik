Это один и тот же адрес! Корчое self это и есть экземпляр класса

```python
# чтобы теперь вызвать метод через сам класс, адрес экземпляра уже автоматически не подставляется. нужно подставить ссылку на этот класс в метод
Point.set_coords(Point)
```

```python
class Point:
    color = 'red'
    circle = 2

    def set_coords(self, x, y):
        self.x = x
        self.y = y

    def get_coords(self):
        return (self.x, self.y)

pt = Point()
pt.set_coords(1, 2)
print(pt.__dict__)
```

```python
pt2 = Point()
pt2.set_coords(10, 20)
```

```python
# благодаря параметру self мы не перепутали, какому объекту какое значение атрибута назначить
print(pt.__dict__)
print(pt2.__dict__)
```

```python
# получили доступ к нашим атрибутам
pt.get_coords()
```

```python
# так как объявленные методы - это тоже атрибуты класса, мы можем получить к ним доступ через getattr
print(getattr(pt, 'get_coords'))
# так как атрибус - это метод (действие), то возвращается функция, и ее можно вызвать через круглые скобки
f = getattr(pt, 'get_coords')
f()
```
